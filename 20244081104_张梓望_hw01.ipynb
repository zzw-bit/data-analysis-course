## 1. 热身：定义简单函数

1. `say_hi()`：无参数，调用时打印 `Hello, world!`
2. `goat_pad(s)`：接受一个字符串，打印该字符串前后各加 `goat`
3. `print_n(s, n)`：接受字符串 `s` 和整数 `n`，把 `s` 打印 `n` 次，每次一行

**思路**：题目要求的是"打印"，所以函数内部用 `print()`，不用 `return`。
字符串用 `+` 拼接；重复打印用 `for _ in range(n)` 循环，`print()` 自带换行。

```python
def say_hi():
    """无参数，打印 Hello, world!"""
    print("Hello, world!")


def goat_pad(s):
    """接受一个字符串，打印 goat + s + goat"""
    print("goat" + s + "goat")


def print_n(s, n):
    """把字符串 s 打印 n 次，每次一行"""
    for _ in range(n):
        print(s)


say_hi()
goat_pad("bird")
print_n("goat", 3)
```

## 2. 欧几里得算法（GCD）

1. 实现 `gcd(m, n)`，接受两个整数，返回最大公约数
2. 计算 (2018, 2019)、(1200, 300)、(5040, 60)
3. 负数参数时的行为与讨论

**思路（辗转相除法）**：不断把 `(m, n)` 替换成 `(n, m % n)`，直到 `n == 0`，此时 `m` 就是 GCD。
函数开头先对两个参数取绝对值，保证返回值非负（数学上 GCD 定义为非负数）。

```python
def gcd(m, n):
    """返回 m 和 n 的最大公约数（欧几里得算法）"""
    m, n = abs(m), abs(n)
    while n != 0:
        m, n = n, m % n
    return m
```

```python
pairs = [(2018, 2019), (1200, 300), (5040, 60)]
for m, n in pairs:
    print(f"gcd({m}, {n}) = {gcd(m, n)}")
```

### 第 3 问：参数为负数时会怎样？

- 如果函数里不做处理，Python 的 `%` 运算结果符号与**除数**一致，算法虽然仍会终止，但可能返回负数，例如 `gcd(-4, 6)` 会得到 `-2`，这与"最大公约数是最大的那个正整数"的直觉不符。
- 上面的实现在开头用 `abs()` 取绝对值，因此 `gcd(-4, 6) == gcd(4, 6) == 2`。
- 这个行为是有意义的：数学上 $\gcd(m,n) = \gcd(|m|,|n|)$，GCD 约定为非负数，取绝对值后结果与手算一致。

```python
print(gcd(-4, 6), gcd(4, -6), gcd(-12, -18))
```

## 3. 近似欧拉数 e

- 极限形式：$e = \lim_{x \to \infty}\left(1+\frac{1}{x}\right)^x$
- 级数形式：$e = \sum_{k=0}^{\infty}\frac{1}{k!}$，约定 $0! = 1$

1. `euler_limit(n)`：返回 $(1+1/n)^n$
2. `euler_infinite_sum(n)`：返回级数前 n 项之和。级数从 k=0 起算，函数从 1 起算：
   `euler_infinite_sum(0)` → 0，`euler_infinite_sum(1)` → 1，`euler_infinite_sum(4)` → 约 2.667
3. `euler_approx(eps)`：用 while 循环累加，直到与 `math.exp(1)` 的误差小于 eps
4. `print_euler_sum_table(n)`：逐行打印 `euler_infinite_sum(k)`，k 从 1 到 n

```python
import math


def euler_limit(n):
    """返回 (1 + 1/n) ** n，作为 e 的近似"""
    return (1 + 1 / n) ** n


print(euler_limit(1), euler_limit(10), euler_limit(100), euler_limit(10000))
print("真实值 e =", math.exp(1))
```

```python
def euler_infinite_sum(n):
    """返回级数 sum_{k=0}^{n-1} 1/k!，n 为非负整数"""
    total = 0.0
    for k in range(n):
        total += 1 / math.factorial(k)
    return float(total)


print(euler_infinite_sum(0))   # 0
print(euler_infinite_sum(1))   # 1.0
print(euler_infinite_sum(4))   # 1 + 1 + 1/2 + 1/6 ≈ 2.667
```

```python
def euler_approx(epsilon):
    """累加 1/k!，直到与真实值 e 的误差小于 epsilon"""
    total = 0.0
    k = 0
    while abs(total - math.exp(1)) >= epsilon:
        total += 1 / math.factorial(k)
        k += 1
    return total


print(euler_approx(1e-1))
print(euler_approx(1e-6))
print(euler_approx(1e-12))
```

```python
def print_euler_sum_table(n):
    """逐行打印 euler_infinite_sum(k)，k 从 1 到 n"""
    for k in range(1, n + 1):
        print(euler_infinite_sum(k))


print_euler_sum_table(10)
```

## 4. 测试整数属性

1. `is_power_of_2(n)`：判断正整数 n 是否是 2 的幂，**不能用 `math.sqrt`**，只用除法和取模
2. `is_power(b, n)`：推广版，判断 n 是否是 b 的幂

**思路**：只要 n 还能被 2 整除就一直除以 2，最后剩下的数若为 1，说明 n 恰好是若干个 2 相乘。
例如 8 → 4 → 2 → 1，剩下 1，是 2 的幂；12 → 6 → 3，剩下 3，不是。
推广时把 2 换成 b 即可，注意 b == 1 时要特判（否则 `n % 1 == 0` 恒真会死循环）。

```python
def is_power_of_2(n):
    """判断正整数 n 是否是 2 的幂（只用 // 和 %）"""
    if n < 1:
        return False
    while n % 2 == 0:
        n //= 2
    return n == 1


print(is_power_of_2(1), is_power_of_2(2), is_power_of_2(8), is_power_of_2(6), is_power_of_2(1024))
```

```python
def is_power(b, n):
    """判断 n 是否是 b 的幂"""
    if n < 1 or b < 1:
        return False
    if b == 1:
        return n == 1
    while n % b == 0:
        n //= b
    return n == 1


print(is_power(2, 8), is_power(3, 9), is_power(3, 8), is_power(5, 125), is_power(1, 1))
```
