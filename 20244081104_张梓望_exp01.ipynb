## 实验 2：函数、条件、循环与递归

### 2.1 score_level(score)

用 `if / elif` 从高到低判断，最后 `else` 兜底；范围外的先抛 `ValueError`。

```python
def score_level(score):
    """90-100 优秀，80-90 良好，60-80 及格，0-60 不及格；超出 0~100 抛 ValueError"""
    if not 0 <= score <= 100:
        raise ValueError(f"成绩必须在 0~100 之间，收到 {score}")
    if score >= 90:
        return "优秀"
    if score >= 80:
        return "良好"
    if score >= 60:
        return "及格"
    return "不及格"


for s in (100, 95, 85, 76, 60, 59, 0):
    print(f"{s} -> {score_level(s)}")

try:
    score_level(120)
except ValueError as e:
    print("捕获 ValueError:", e)
```

### 2.2 for 与 while 求 1~100 的和

```python
total_for = 0
for i in range(1, 101):
    total_for += i

total_while = 0
i = 1
while i <= 100:
    total_while += i
    i += 1

print("for   :", total_for)
print("while :", total_while)
print("两者相等:", total_for == total_while)
```

### 2.3 递归与迭代阶乘

递归的两个要素：**基线条件**（n == 0 返回 1）+ **递归条件**（n * fact(n-1)）。

```python
def fact_recursive(n):
    """递归版阶乘，0! = 1"""
    if n < 0:
        raise ValueError("n 必须是非负整数")
    if n == 0:
        return 1
    return n * fact_recursive(n - 1)


def fact_iterative(n):
    """迭代版阶乘，0! = 1"""
    if n < 0:
        raise ValueError("n 必须是非负整数")
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result


for n in (0, 1, 5, 10):
    print(f"{n}! 递归 = {fact_recursive(n):>8}  迭代 = {fact_iterative(n):>8}")
```

### 2.4 summary(values) 与元组解包

```python
def summary(values):
    """非空数值列表 -> 元组 (最小值, 最大值, 平均值)"""
    if not values:
        raise ValueError("values 不能为空")
    return min(values), max(values), sum(values) / len(values)


lowest, highest, average = summary([85, 58, 92, 76, 60, 89])
print("最小值:", lowest)
print("最大值:", highest)
print("平均值:", average)
```

### 2.5 sum_all(*args)

`*args` 把任意多个位置参数收集成一个元组。

```python
def sum_all(*args):
    """返回所有位置参数之和"""
    return sum(args)


print("无参数 :", sum_all())
print("单参数 :", sum_all(5))
print("多参数 :", sum_all(1, 2, 3, 4, 5))
```

### 4.3 编写 parse_student(line) 并保存为模块 student_utils.py

校验顺序：字段数量 → 姓名非空 → 成绩能转成整数 → 成绩在 0~100。任一不满足抛 `ValueError`。

```python
%%writefile student_utils.py
# -*- coding: utf-8 -*-
"""学生记录解析模块。"""


def parse_student(line):
    """把一行 CSV 文本解析为 {"name": str, "score": int}。无效记录抛 ValueError。"""
    parts = line.rstrip("\n").split(",")
    if len(parts) != 2:
        raise ValueError(f"字段数量应为 2，实际为 {len(parts)}：{line!r}")

    name, raw_score = parts[0].strip(), parts[1].strip()
    if not name:
        raise ValueError(f"姓名不能为空：{line!r}")

    try:
        score = int(raw_score)
    except ValueError:
        raise ValueError(f"成绩不是整数：{raw_score!r}") from None

    if not 0 <= score <= 100:
        raise ValueError(f"成绩超出 0~100 范围：{score}")

    return {"name": name, "score": score}
```

### 4.4 逐行处理，捕获异常并记录行号

`enumerate(f, start=1)` 直接给出文件行号；遇到非法记录只记录，不中断循环。

```python
import importlib
import student_utils
importlib.reload(student_utils)  # 保证用的是刚写入的最新版本

valid_records = []
invalid_records = []

with open(FILE_PATH, encoding="utf-8") as f:
    for line_no, line in enumerate(f, start=1):
        if line_no == 1 or not line.strip():
            continue  # 跳过表头与空行
        try:
            valid_records.append(student_utils.parse_student(line))
        except ValueError as e:
            invalid_records.append((line_no, str(e)))

print("有效记录:", valid_records)
print("无效记录:")
for line_no, reason in invalid_records:
    print(f"  第 {line_no} 行：{reason}")
```

## 实验 5：类、继承与对象复制

### 5.1 / 5.2 Student 类与实例化

`__init__()` 初始化属性，`__str__()` 决定 `print(对象)` 显示什么。

```python
class Student:
    """学生类：保存姓名与成绩"""

    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        """成绩是否及格"""
        return self.score >= 60

    def __str__(self):
        return f"Student(name={self.name}, score={self.score}, passed={self.is_passed()})"


students = [Student(r["name"], r["score"]) for r in valid_records]
for stu in students:
    print(stu, "| is_passed():", stu.is_passed())
```

### 5.3 GraduateStudent 继承

`super().__init__()` 复用父类初始化逻辑，子类只多存一个 `research_topic`，并重写 `__str__()`。

```python
class GraduateStudent(Student):
    """继承 Student，增加 research_topic"""

    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        return (f"GraduateStudent(name={self.name}, score={self.score}, "
                f"research_topic={self.research_topic}, passed={self.is_passed()})")


gs = GraduateStudent("Alice", 92, "Data Mining")
print(gs)
print("继承自父类的方法 is_passed():", gs.is_passed())
print("是否是 Student 的实例:", isinstance(gs, Student))
```

### 5.4 直接赋值 / copy.copy() / copy.deepcopy()

- 直接赋值：完全同一个对象，改一处两处都变
- `copy.copy()`：浅拷贝，只复制最外层，内部的嵌套列表仍共享
- `copy.deepcopy()`：深拷贝，内部嵌套对象也复制，互不影响

每次比较都从新建的原始对象开始。

```python
import copy


def compare(kind):
    """按指定方式复制对象并修改内部列表，返回 (原始对象, 复制对象)"""
    original = {"name": "Alice", "scores": [85, 90]}
    if kind == "直接赋值":
        obj = original
    elif kind == "浅拷贝":
        obj = copy.copy(original)
    else:
        obj = copy.deepcopy(original)

    obj["scores"].append(100)
    obj["name"] = "Bob"
    return original, obj


for kind in ("直接赋值", "浅拷贝", "深拷贝"):
    original, modified = compare(kind)
    print(f"--- {kind} ---")
    print("  原始对象:", original)
    print("  修改后  :", modified)
```
